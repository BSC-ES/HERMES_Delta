# How to create rotated grids

```python
import nes
import matplotlib
import matplotlib.pyplot as plt
import geopandas as gpd
```

```python
%matplotlib inline
```

## 1. Create dataset

### Define grid details

This corresponds to a regular grid for CAMS2_40 forecast

```python
projection='rotated'
centre_lat = 51
centre_lon = 10
west_boundary = -35
south_boundary = -27
inc_rlat = 0.15
inc_rlon = 0.15
```

```python
nessy = nes.create_nes(comm=None, info=False, projection=projection,
                       centre_lat=centre_lat, centre_lon=centre_lon,
                       west_boundary=west_boundary, south_boundary=south_boundary,
                       inc_rlat=inc_rlat, inc_rlon=inc_rlon)
```

### Read projection

```python
nessy.projection
```

```python
nessy.projection_data
```

## 2. Plot

```python
nessy.create_shapefile()
```

```python
fig, ax = plt.subplots(1, figsize=(19, 7))
nessy.shapefile.plot(ax=ax, facecolor="blue", edgecolor="blue")
countries = gpd.read_file('/gpfs/projects/bsc32/models/NES_tutorial_data/gadm_country_mask/gadm_country_ISO3166.shp')
countries.plot(ax=ax, facecolor="none", edgecolor='black', linewidth=0.3)
ax.margins(0)
```

## 3. Write dataset

### Write NetCDF

```python
nessy.to_netcdf('rotated_grid.nc', info=True)
```

### Write shapefile

```python
nessy.to_shapefile('rotated_grid_shp')
```
